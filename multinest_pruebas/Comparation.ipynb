## FlatLambdaCDM

Primero cargamos las cadenas y usamos solo los ultimos 30000 puntos de la cadena.

```python
chain_flcdm = np.genfromtxt("/Users/juda/lab_alma/multinest_pruebas/chain_supernova_flcdm/ev.dat", usemask=False, usecols=(0,1,2,3))
```

```python
labels = ['alpha', 'beta', 'H0', 'Om', 'Ol', 'w0']
fig = corner.corner(
    chain_flcdm[30000:], labels=labels, show_titles=True);
```

Podemos apreciar que se forman histogramas que parecen ser gausianos y contornos circulares excepcion del contorno formado por H0 y Om, que tiene forma eliptica, pero sin presentar una degeneración importante.

## LambdaCDM

De igual forma cargamos las cadenas y usamos solo los ultimos 30000 puntos de la cadena.

```python
chain_lcdm = np.genfromtxt("/Users/juda/lab_alma/multinest_pruebas/chain_supernova_lcdm/ev.dat", usemask=False, usecols=(0,1,2,3,4))
```

```python
fig = corner.corner(
    chain_lcdm[30000:], labels=labels, show_titles=True);
```

En este punto de la cadena los puntos parecen ya haber convergido y la mayoria no presenta degeneraciones, siendo el que mas presenta el contorno formado por Ol y Om, esto puede ser explicado al pensar que la suma de las Omega debe estar acotada, al dejar libre la curvatura, la suma de Ol y Om esta mas libre pero sigue estando correlacionada a tomar valores que reproduzcan que la suma de todas las Omega debe ser 1.
