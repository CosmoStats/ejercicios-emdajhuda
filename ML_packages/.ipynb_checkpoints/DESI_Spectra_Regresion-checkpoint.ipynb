```python
#Preparar datos para la regresión
X=[]
Y=[]
for i in range(len(targetids)):
  spectype = df_spectra.loc[df_spectra['targetid'] == targetids[i], 'spectype'].values[0]
  #redshift = df_spectra.loc[df_spectra['targetid'] == targetids[i], 'z'].values[0]
  #Y.append(df_spectra.loc[df_spectra['targetid'] == targetids[i], 'z'].values[0])
  if spectype:
    if spectype == 'GALAXY':
       Y.append(df_spectra.loc[df_spectra['targetid'] == targetids[i], 'z'].values[0])
  #  elif spectype == 'STAR':
  #    Y.append(1)
  #  elif spectype == 'QSO':
  #    Y.append(2)
  #  else:
  #    print(f"spectype for TID:{spectype} not accepted")
  #    continue
  #else:
  #  print(f"No spectype for TID:{targetids[i]}")
  #  continue
       X.append(data['flux_arrays'][i])

X=np.asarray(X)
Y=np.asarray(Y)
```

```python
X.shape,Y.shape
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Split into train and test sets
X_train, X_test, y_train, y_test = train_test_split(
X, Y, test_size=0.3, random_state=112)
```

```python
X_train.shape
```

```python
original_shape = X_train.shape[1:]
X_train_flat = X_train.reshape(X_train.shape[0], -1)
X_test_flat = X_test.reshape(X_test.shape[0], -1)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_flat)
X_test_scaled = scaler.transform(X_test_flat)

# Reshape back to original format
X_train = X_train_scaled.reshape(X_train.shape[0], *original_shape)
X_test = X_test_scaled.reshape(X_test.shape[0], *original_shape)
X_train.shape
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
```

```python
#Train NN for classification of spectra
model = MLPRegressor(hidden_layer_sizes=(200,120, 70, 20),
                                     random_state=42, max_iter=1000)
#model = MLPRegressor(hidden_layer_sizes=(100, 50, 30),
  #                                   random_state=42, max_iter=1000)
#model=MLPClassifier(hidden_layer_sizes=(100, 50, 30),
#                                      random_state=42, max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Calculate regression metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)

# Print summary of results
print("\n" + "="*60)
print("REGRESSION MODEL COMPARISON SUMMARY")
print("="*60)
print(f"R2 = {r2:.4f}, MSE = {mse:.4f}, MAE = {mae:.4f}")
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import mean_squared_error, r2_score

# Create a comparison plot
fig, axes = plt.subplots(1, 1, figsize=(15, 12))
#xes = axes.ravel()

# Scatter plot of predicted vs actual
axes.scatter(y_test, y_pred, alpha=0.6, s=30, label='Predictions')

# Add perfect prediction line (diagonal)
min_val = 0
max_val = 5
#axes.plot([min_val, max_val], [min_val, max_val], 'r--', linewidth=2, label='Perfect Prediction')

axes.set_xlabel('True Redshift', fontsize=12)
axes.set_ylabel('Predicted Redshift', fontsize=12)
axes.set_title(f'$R^2$ = {r2:.4f}, MSE = {mse:.4f}', fontsize=12)
axes.legend()
axes.grid(True, alpha=0.3)
axes.axis('equal')
plt.xlim(min_val,max_val)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
# Create a comparison plot
fig, axes = plt.subplots(1, 1, figsize=(15, 12))

# Create a comparison plot
cm = confusion_matrix(y_test,y_pred, normalize='true')  # Normalize by true labels

# Display confusion matrix
disp = ConfusionMatrixDisplay(confusion_matrix=cm,
                                display_labels=['GALAXY', 'STAR', 'QSO'])  # Use class names for better readability
disp.plot(ax=axes, cmap='Blues', values_format='.2f')  # Format values to 2 decimal places

plt.tight_layout()
plt.show()
```
